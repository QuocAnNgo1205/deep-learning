```python
import numpy as np
import matplotlib.pyplot as plt

# Khởi tạo số lượng hình tròn
n_circles = 50

# 1. Sinh toạ độ tâm (x, y) ngẫu nhiên trong khoảng [0, 1]
x_coords = np.random.rand(n_circles)
y_coords = np.random.rand(n_circles)

# 2. Sinh màu sắc ngẫu nhiên trong khoảng [0, 1]
colors = np.random.rand(n_circles)

# 3. Sinh bán kính r trong khoảng [0, 1] và tính diện tích
radii = np.random.rand(n_circles)
areas = np.pi * (radii ** 2)

# ==========================================
# PHẦN HIỂN THỊ ĐỒ THỊ
# ==========================================
fig, ax = plt.subplots(figsize=(7, 7))

# scale_factor giúp hình tròn to ra trên đồ thị để dễ quan sát (không làm đổi dữ liệu gốc)
scale_factor = 400 

ax.scatter(x_coords, y_coords, 
           s=areas * scale_factor, 
           c=colors, 
           alpha=0.6, 
           edgecolors='white', 
           linewidth=0.5)

ax.set_title("Biểu đồ phân bố 50 hình tròn ngẫu nhiên")
ax.set_xlabel("Tọa độ X")
ax.set_ylabel("Tọa độ Y")

# Giới hạn trục toạ độ đúng trong khoảng [0, 1]
ax.set_xlim(0, 1)
ax.set_ylim(0, 1)
ax.grid(True, linestyle='--', alpha=0.3)

plt.show()
```
