```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Sinh dữ liệu trong khoảng [-pi, pi]
x_vals = np.linspace(-np.pi, np.pi, 256, endpoint=True)
sin_vals = np.sin(x_vals)
cos_vals = np.cos(x_vals)

# 2. Thiết lập khung đồ thị
plt.figure(figsize=(8, 5))

# Vẽ hai hàm số với màu và nét đứt khác nhau để dễ phân biệt
plt.plot(x_vals, sin_vals, color="red", linewidth=2, linestyle="-", label=r'$\sin(x)$')
plt.plot(x_vals, cos_vals, color="blue", linewidth=2, linestyle="--", label=r'$\cos(x)$')

# 3. Định dạng các trục toạ độ
plt.xlim(x_vals.min() * 1.1, x_vals.max() * 1.1)
plt.ylim(-1.1, 1.1)

# Thay thế các giá trị số bằng ký hiệu toán học trên trục
plt.xticks([-np.pi, -np.pi / 2, 0, np.pi / 2, np.pi],
           [r'$-\pi$', r'$-\pi/2$', r'$0$', r'$+\pi/2$', r'$+\pi$'])
plt.yticks([-1, 0, 1], [r'$-1$', r'$0$', r'$+1$'])

# 4. Thêm chi tiết phụ trợ
plt.title("Đồ thị hàm số lượng giác trong khoảng $[-\pi, \pi]$")
plt.legend(loc="upper left") # Hiển thị chú giải
plt.grid(True, linestyle=':', alpha=0.7) # Thêm lưới mờ

plt.show()
```
